# 🟢 Easy: GELU Activation

Implement the **GELU** (Gaussian Error Linear Unit) activation.

$$\text{GELU}(x) = x \cdot \Phi(x) = x \cdot 0.5 \cdot (1 + \text{erf}(x / \sqrt{2}))$$

### Signature
```python
def my_gelu(x: Tensor) -> Tensor: ...
```

### Rules
- Do NOT use `F.gelu`, `nn.GELU`, or `torch.nn.functional.gelu`
- Use `torch.erf` for the exact version

````python
import torch
import math
````

````python
# ✏️ YOUR IMPLEMENTATION HERE

def my_gelu(x):
    return 0.5 * x * (1 + torch.erf(x / math.sqrt(2)))
````

````python
# 🧪 Debug
x = torch.tensor([-2., -1., 0., 1., 2.])
print('Output:', my_gelu(x))
print('Ref:   ', torch.nn.functional.gelu(x))
````

````python
# ✅ SUBMIT
from torch_judge import check
check('gelu')
````

````python

import torch.nn.functional as F

# torch.linsapce: linearly spaced vectors 생성
x = torch.linspace(-5, 5, 1000)

y_custom = my_gelu(x)
y_pytorch = F.gelu(x, approximate="none")

print("All close:",
      torch.allclose(y_custom, y_pytorch, atol=1e-6))

print("Maximum error:",
      (y_custom - y_pytorch).abs().max().item())
````

````python
x = torch.tensor([1., 2.], requires_grad=True)

y = my_gelu(x)
loss = y.sum()

loss.backward()

print(x.grad)
````
